## Why Conditionals Matter

Conditional statements let a program **make decisions** — running different code depending on whether an expression evaluates to `True` or `False`. Python relies on **indentation** (not braces `{}`) to define which statements belong to a block, so consistent whitespace is not optional — it's syntax.


---

## 1. The `if` Statement

Executes a block of code **only if** the condition is `True`. If the condition is `False`, the block is simply skipped.


````python
age = 20

if age >= 18:
    print("Eligible to vote.")
````

### Short-Hand `if`

If the body is a single statement, it can be written on the same line as the `if`. Handy for quick checks, but avoid it once logic gets more complex — readability drops fast.


````python
age = 19
if age > 18: print("Eligible to Vote.")
````

---

## 2. The `if-else` Statement

Runs **one** block when the condition is `True` and a **different** block when it's `False`. Exactly one of the two branches always executes.


````python
age = 10

if age <= 12:
    print("Travel for free.")
else:
    print("Pay for ticket.")
````

---

## 3. The `if-elif-else` Statement

Used to check **multiple, mutually exclusive** conditions in sequence. Python evaluates each `elif` top-to-bottom and stops at the **first** one that's `True` — later conditions are never checked, even if they'd also be `True`.


````python
age = 25

if age <= 12:
    print("Child.")
elif age <= 19:
    print("Teenager.")
elif age <= 35:
    print("Young adult.")
else:
    print("Adult.")
````

**Explanation:** Since `age` is `25`, it skips `age <= 12` and `age <= 19` (both `False`), then `age <= 35` is `True`, so `"Young adult."` prints and the chain stops there — `else` never runs.

> **Common pitfall:** Order matters. If you wrote `elif age <= 100` *before* `elif age <= 35`, the more specific check after it would never be reached. Always put the more restrictive conditions first.


---

## 4. Nested `if-else` Statements

An `if`/`else` placed **inside** another `if` or `else` block — used to check a condition only after another condition has already been satisfied.


````python
age = 70
is_member = True

if age >= 60:
    if is_member:
        print("30% senior discount!")
    else:
        print("20% senior discount.")
else:
    print("Not eligible for a senior discount.")
````

> **Tip:** Nesting more than 2–3 levels deep usually hurts readability. Often you can flatten nested conditions using `and`:
> ```python
> if age >= 60 and is_member:
>     print("30% senior discount!")
> ```


---

## 5. Conditional Expression (Ternary Operator)

A compact, single-line way to assign one of two values based on a condition. Syntax: `value_if_true if condition else value_if_false`.


````python
age = 20
s = "Adult" if age >= 18 else "Minor"
print(s)
````

**Explanation:** If `age >= 18` is `True`, `s` is assigned `"Adult"`. Otherwise, it's assigned `"Minor"`.

> **Style note:** Ternary expressions are great for simple value selection, but chaining several together (`a if x else b if y else c`) quickly becomes hard to read — prefer a full `if-elif-else` once logic grows.


---

## 6. The `match-case` Statement (Python 3.10+)

Compares a value against multiple **patterns**, similar to `switch-case` in other languages. `|` matches multiple values with one case, and `_` is the wildcard/default case.


````python
number = 2

match number:
    case 1:
        print("One")
    case 2 | 3:
        print("Two or Three")
    case _:
        print("Other number")
````

`match-case` supports more than literal matching — it can also destructure sequences, dicts, and objects:


````python
def describe(point):
    match point:
        case (0, 0):
            return "Origin"
        case (x, 0):
            return f"On the X-axis at {x}"
        case (0, y):
            return f"On the Y-axis at {y}"
        case (x, y):
            return f"Point at ({x}, {y})"
        case _:
            return "Not a point"

print(describe((0, 0)))
print(describe((5, 0)))
print(describe((3, 4)))
````

## 7. Quick Reference

| Construct | Use When |
|---|---|
| `if` | Run code only for one condition |
| `if / else` | Exactly two mutually exclusive outcomes |
| `if / elif / else` | Several mutually exclusive conditions |
| Nested `if` | A condition only matters after another is met |
| Ternary (`x if c else y`) | Simple single-value selection, inline |
| `match-case` | Comparing one value against many known patterns/shapes |

## 8. Common Pitfalls

- **Using `=` instead of `==`** — `if age = 18:` is a `SyntaxError` in Python (good — it protects you from the classic C bug), but always double-check comparisons.
- **Forgetting indentation** — Python uses indentation to define blocks; mixing tabs and spaces raises `IndentationError`.
- **Chained comparisons** — Python lets you write `0 < age < 18` directly instead of `age > 0 and age < 18`. It's valid and often clearer.
- **Truthy/falsy shortcuts** — `if my_list:` checks whether the list is non-empty; you don't need `if len(my_list) > 0:`.


````python
age = 15
# chained comparison — valid and idiomatic in Python
print(0 < age < 18)

my_list = []
if my_list:
    print("has items")
else:
    print("empty list")
````
