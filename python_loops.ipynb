## Why Loops Matter

Loops let a program **execute a block of code repeatedly**, either for every item in a sequence or until a condition stops being true. Python has two loop constructs:

- **`for` loop** — iterate over a sequence (list, tuple, string, `range`, etc.)
- **`while` loop** — repeat based on a condition


---

## 1. The `for` Loop

Executes a block of code once for **each item** in a sequence.


````python
n = 4
for i in range(0, n):
    print(i)
````

**Explanation:**
- `n = 4` stores the ending value for the loop.
- `range(0, n)` generates numbers from `0` to `3` (the stop value `n` is **excluded**).
- `for i in range(0, n):` iterates through each number in the range.

> **Note:** `range(n)` is shorthand for `range(0, n)`. `range()` also accepts a step: `range(0, 10, 2)` → `0, 2, 4, 6, 8`.


````python
# range with a step, and iterating directly over other sequences
for i in range(0, 10, 2):
    print(i, end=' ')
print()

for ch in "geeks":
    print(ch, end='-')
print()

for fruit in ("apple", "banana", "cherry"):
    print(fruit)
````

### Iterating by Index of Sequences

A `for` loop can iterate through elements **using their index**, with the help of `range()` and `len()`.


````python
a = ["geeks", "for", "geeks"]
for idx in range(len(a)):
    print(a[idx])
````

**Explanation:**
- `len(a)` returns the total number of elements in the list.
- `range(len(a))` generates index values from `0` to `2`.
- `a[idx]` accesses list elements using their index.

> **Pythonic tip:** If you only need the *values*, loop directly (`for item in a:`) — it's cleaner. Use index-based iteration only when you actually need the index, or better yet, use `enumerate()`:


````python
a = ["geeks", "for", "geeks"]
for idx, value in enumerate(a):
    print(idx, "->", value)
````

## 2. The `while` Loop

Repeatedly executes a block of code **as long as** the given condition remains `True`. Once the condition becomes `False`, execution moves to the line right after the loop.


````python
cnt = 0
while (cnt < 3):
    cnt = cnt + 1
    print("Hello Geek")
````

**Explanation:**
- `cnt = 0` initializes the counter variable.
- `while (cnt < 3):` runs the loop while the condition is true.
- `cnt = cnt + 1` increases the counter value by 1.

> **Watch out:** Forgetting to update the loop variable (`cnt = cnt + 1` here) inside a `while` loop is the #1 cause of accidental infinite loops.


### Infinite `while` Loop

An infinite loop runs continuously because its condition always remains `True`.

```python
while (True):
    print("Hello Geek")
```

- `while(True):` creates a loop with a condition that is always true.
- `print("Hello Geek")` keeps executing repeatedly inside the loop.

> It's not recommended to run this as-is since it never stops on its own — you'd have to forcefully interrupt the program. In practice, infinite loops are used intentionally with a `break` condition inside, e.g. servers or event loops waiting for input.


````python
# A *controlled* "infinite" loop — the common real-world pattern
count = 0
while True:
    print("Hello Geek")
    count += 1
    if count == 3:
        break   # exits the loop, avoiding a true infinite run
````

## 3. Nested Loops

A loop **inside** another loop. The inner loop runs completely for every single iteration of the outer loop.


````python
for i in range(1, 5):
    for j in range(i):
        print(i, end=' ')
    print()
````

**Explanation:**
- The outer loop `for i in range(1, 5)` controls the number of rows.
- The inner loop `for j in range(i)` runs `i` times for each row.
- `print(i, end=' ')` prints the value of `i` on the same line; `print()` moves to a new line after each row.

> **Complexity note:** Nested loops multiply work — an outer loop of `n` iterations with an inner loop of `m` iterations does `n * m` total iterations. Watch this when looping over large datasets.

## 4. Loop Control Statements

Python provides three statements to change a loop's normal flow:

| Statement | Effect |
|---|---|
| `break` | Exits the loop immediately, skipping any remaining iterations |
| `continue` | Skips the rest of the current iteration and moves to the next one |
| `pass` | Does nothing — a placeholder where a statement is syntactically required |


````python
# break — stop as soon as we find what we need
for num in range(10):
    if num == 5:
        break
    print(num, end=' ')
print("\n-- loop stopped early via break --")
````

````python
# continue — skip even numbers, print only odd ones
for num in range(10):
    if num % 2 == 0:
        continue
    print(num, end=' ')
print("\n-- even numbers skipped via continue --")
````

````python
# pass — a syntactic no-op, useful as a placeholder while writing code
for num in range(5):
    if num == 3:
        pass  # TODO: handle this case later
    else:
        print(num, end=' ')
print()
````

### The `else` Clause on Loops

Python's `for` and `while` loops can have an `else` block — it runs **only if the loop completes without hitting a `break`**. This is a Python-specific feature not found in most languages.


````python
# else runs because the loop finished naturally (no break)
for i in range(3):
    print(i)
else:
    print("Loop finished without break")

print("---")

# else is skipped because break was hit
for i in range(3):
    if i == 1:
        break
    print(i)
else:
    print("This will NOT print")
````

## 5. `for` vs `while` — When to Use Which

| Use `for` when... | Use `while` when... |
|---|---|
| You know the sequence/range to iterate over | You don't know in advance how many iterations are needed |
| Looping over a list, string, dict, file, etc. | Looping until some condition changes (e.g. user input, sensor reading) |
| Fixed number of repetitions | Open-ended repetition |

## 6. Common Pitfalls

- **Off-by-one errors** — remember `range(n)` stops *before* `n`, not at it.
- **Forgetting to update the condition variable** in a `while` loop → infinite loop.
- **Mutating a list while iterating over it** — can skip elements or cause unexpected behavior. Iterate over a copy (`for x in my_list[:]:`) if you need to modify the original during the loop.
- **Confusing `break` with `continue`** — `break` exits the whole loop; `continue` only skips to the next iteration.


````python
# Pitfall demo: mutating a list while iterating over it
nums = [1, 2, 3, 4, 5]

# BAD: modifying the list you're iterating over can skip elements
# for n in nums:
#     if n % 2 == 0:
#         nums.remove(n)

# GOOD: iterate over a copy instead
for n in nums[:]:
    if n % 2 == 0:
        nums.remove(n)

print(nums)  # [1, 3, 5]
````
