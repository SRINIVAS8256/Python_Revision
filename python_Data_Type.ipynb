## Why Data Types Matter

In Python, **every value is an object**, and every object has a type. The type determines:

- **How the value is stored** in memory
- **What operations** are valid on it (`+`, indexing, iteration, etc.)
- **Whether it can be changed** after creation (mutability)

Python is **dynamically typed** — you don't declare a type; the interpreter infers it from the value assigned, and a variable can be reassigned to a completely different type at any time.


```python
# A variable's type changes as its value changes
x = 50
print(type(x))

x = 60.5
print(type(x))

x = "Hello World"
print(type(x))

x = ["geeks", "for", "geeks"]
print(type(x))

x = ("geeks", "for", "geeks")
print(type(x))
```

## 1. Numeric Data Types

Python has three built-in numeric types:

| Type | Class | Description | Example |
|---|---|---|---|
| Integer | `int` | Whole numbers, positive or negative, **no size limit** | `-7`, `0`, `42` |
| Float | `float` | Real numbers with a decimal point | `3.14`, `-0.5` |
| Complex | `complex` | Numbers with real + imaginary parts | `2 + 4j` |

> **Pro tip:** Unlike languages such as C or Java, Python's `int` has arbitrary precision — it grows automatically to hold numbers of any size, limited only by available memory.


```python
a = 5
b = 5.0
c = 2 + 4j

print(type(a), a)
print(type(b), b)
print(type(c), c)

# complex numbers expose .real and .imag
print("Real part:", c.real)
print("Imaginary part:", c.imag)

# int has no fixed size — this works fine
big_number = 2 ** 200
print(big_number)
```

## 2. Sequence Data Types

A **sequence** is an ordered collection where each element has a fixed position (index). Sequences support:
- Indexing (`s[0]`, `s[-1]`)
- Slicing (`s[1:3]`)
- Iteration (`for item in s`)
- `len()`

### 2.1 String (`str`)

Strings store text and are **immutable** — once created, individual characters cannot be reassigned.


```python
s = 'Welcome to the Geeks World'
print(s)
print(type(s))

# Indexing
print(s[1])    # second character
print(s[-1])   # last character

# Slicing
print(s[0:7])  # 'Welcome'

# Strings are immutable — this raises TypeError:
try:
    s[0] = 'w'
except TypeError as e:
    print("Error:", e)
```

### 2.2 List (`list`)

Lists are **ordered and mutable** — you can add, remove, or change elements after creation. They can hold mixed data types.


```python
a = [1, 2, 3]
print(a)

b = ["Geeks", "For", "Geeks", 4, 5]
print(b[3])    # 4
print(b[-3])   # 'Geeks'

# Mutability in action
b.append("Practice")
b[0] = "geeksforgeeks"
print(b)
```

### 2.3 Tuple (`tuple`)

Tuples are **ordered and immutable**. They're commonly used for fixed collections of data (e.g. coordinates) and are slightly faster and more memory-efficient than lists.

> **Gotcha:** A single-element tuple *must* have a trailing comma — `(1,)` is a tuple, but `(1)` is just the integer `1` in parentheses.


```python
t1 = (1,)
print(type(t1))   # <class 'tuple'>

not_a_tuple = (1)
print(type(not_a_tuple))  # <class 'int'>

t2 = ('Geeks', 'For', 'Geeks', 1, 2)
print(t2[3])    # 1
print(t2[-3])   # 'Geeks'

# Tuples are immutable — this raises TypeError:
try:
    t2[0] = 'X'
except TypeError as e:
    print("Error:", e)
```

## 3. Boolean Data Type (`bool`)

Booleans represent `True` or `False`. Internally, `bool` is a **subclass of `int`** — `True == 1` and `False == 0`.


```python
print(type(True))
print(type(False))

# bool is a subclass of int
print(isinstance(True, int))   # True
print(True + True)             # 2
print(True == 1, False == 0)
```

### Truthy and Falsy Values

Every object in Python has an inherent truth value used in conditions (`if`, `while`, etc.).

**Falsy values include:** `0`, `0.0`, `""`, `[]`, `()`, `{}`, `set()`, `None`, `False`
**Everything else is truthy.**


```python
if 1:
    print("1 is truthy")

if not 0:
    print("0 is falsy")

# more examples
for value in [0, 1, "", "text", [], [0], None, {}]:
    print(f"{value!r:10} -> {'truthy' if value else 'falsy'}")
```

## 4. Set Data Type (`set`)

Sets are **unordered, mutable collections of unique elements**. Duplicates are automatically removed. Because sets are unordered, elements aren't accessed by index — you iterate over them instead.


```python
s1 = {"a", "a", "b", "c", "b"}
print(s1)   # duplicates removed automatically

s2 = {"Geeks", "For", "Geeks"}
for i in s2:
    print(i)

# common set operations
odds = {1, 3, 5, 7}
evens = {2, 4, 6, 8}
print(odds | evens)   # union
print(odds & {3, 4, 5})  # intersection
```

> **Note:** Set order is not guaranteed and may vary between runs — never rely on it.

## 5. Dictionary Data Type (`dict`)

Dictionaries store data as **key : value** pairs. Keys must be unique and hashable (e.g. strings, numbers, tuples); values can be anything.

> **Note:** Dictionary keys are case-sensitive — `"Key"` and `"key"` are treated as two different keys.


```python
d = {1: 'Geeks', 2: 'For', 3: 'Geeks'}
print(d[1])          # square-bracket access
print(d.get(2))       # safer access with .get()

# .get() avoids a KeyError for missing keys
print(d.get(99, "Not found"))

# case sensitivity
d2 = {"Key": 1, "key": 2}
print(d2)
```

## 7. Checking and Converting Types

Two useful built-ins for working with types:
- `type(x)` — returns the exact type of `x`
- `isinstance(x, T)` — checks if `x` is an instance of `T` (or a subclass), the preferred way to check types in real code


```python
values = [42, 3.14, "hi", [1, 2], (1, 2), {1, 2}, {"a": 1}, True, None]

for v in values:
    print(f"{str(v):15} -> type: {type(v).__name__:8} | isinstance int? {isinstance(v, int)}")
```

```python
# Type conversion (casting)
print(int("100"))       # str -> int
print(float("3.14"))    # str -> float
print(str(100))         # int -> str
print(list("abc"))      # str -> list of chars
print(tuple([1, 2, 3])) # list -> tuple
print(set([1, 1, 2, 3]))# list -> set (dedupes)
```
